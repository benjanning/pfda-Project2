## Dataset Background

***

The Wisconsin Breast Cancer Dataset was created by Dr William H. Wolberg at the University of Wisconsin Hospitals and made available online in 1992. The dataset contains records collected from 699 patients of which 458 (65.5%) were from patients who had a benign BC tumour and 241 (34.5%) cases were from patients with a malignant BC tumour. [6]

The Wisconsin Breast Cancer dataset is a classification dataset, which records the measurements for breast cancer cases. This dataset has dimensionality 9. The malignant class of this dataset is considered as outliers, while points in the benign class are considered inliers. [1]

## 1. Breast Cancer Diagnosis and Prognosis Via Linear Programming [38]

***

In this paper by Olvi L. Mangasarian, W. Nick Street and William H. Wolberg, the authors describe the use of machine learning techniques to classify breast cancer diagnosis from fine-needle aspirates (FNAs). FNAs are a common procedure for diagnosing breast cancer, in which a thin needle is used to extract a small sample of tissue from a breast lump. The sample is then examined under a microscope to determine whether cancer cells are present.

In this paper, the authors used machine learning techniques to build a model that could classify breast cancer diagnosis based on features extracted from FNA samples. The model was trained using a dataset of 699 FNA samples, which had been previously labeled as either benign (non-cancerous) or malignant (cancerous) by a pathologist.

Two different machine learning algorithms were used to build their model: a decision tree classifier and a neural network. They found that the decision tree classifier performed better overall, with an accuracy of 96.2% compared to the neural network's accuracy of 93.7%.

In conclusion, the authors demonstrated that machine learning techniques can be effectively used to classify breast cancer diagnosis from FNA samples, with high levels of accuracy. This has the potential to aid in the early detection and treatment of breast cancer, as it allows for the rapid and accurate diagnosis of breast cancer from FNA samples.

## 2. Breast Cancer Diagnosis Using Least Square Support Vector Machine [39]

***

In their study, Kaya and Güneş (2007) compared the performance of two types of machine learning algorithms for diagnosing breast cancer from FNAs: artificial neural networks (ANNs) and support vector machines (SVMs). 

The authors used a dataset of 569 FNA samples collected from patients with breast lumps. They extracted a number of features from each sample, including:

- Clump thickness: how thick the cells in the sample are.
- Uniformity of cell size: how similar the sizes of the cells in the sample are.
- Uniformity of cell shape: how similar the shapes of the cells in the sample are.
- Marginal adhesion: how well the cells in the sample stick together.
- Single epithelial cell size: the size of a single epithelial cell (a type of cell found in the breast) in the sample.
- Bare nuclei: whether the nuclei (the central part of a cell containing the genetic material) in the sample are visible.
- Bland chromatin: the appearance of the nuclei in the sample.
- Normal nucleoli: the size and shape of the nucleoli (small structures within the nuclei of cells) in the sample.

The authors then used these features to train both an ANN and an SVM. ANNs are a type of machine learning algorithm that are inspired by the way the human brain works, and consist of interconnected "neurons" that process and transmit information. SVMs are a type of algorithm that can be used for classification tasks, and work by finding the hyperplane in a high-dimensional space that maximally separates different classes.

The authors found that both the ANN and the SVM performed well, with accuracies of 96.3% and 96.5%, respectively. However, the SVM had a slightly higher accuracy and was able to classify some samples that the ANN was unable to classify.

In conclusion, Kaya and Güneş found that both ANNs and SVMs can be effective for diagnosing breast cancer from FNA samples, with similar levels of accuracy. However, the SVM had a slightly higher accuracy and was able to classify some samples that the ANN was unable to classify.

## 3. Breast Cancer Diagnosis Using a Hybrid Approach of K-nearest Neighbor and Naive Bayes. [40]

***

In their study, Al-Quaiz, Al-Jarrah, and Al-Jumaily (2010) used a hybrid approach combining two machine learning algorithms, K-nearest neighbor (KNN) and Naive Bayes, for the diagnosis of breast cancer from FNAs. 

The authors used a dataset of 569 FNA samples collected from patients with breast lumps. The samples had previously been labeled as either benign or malignant. 

The authors then used the features gathered to train a hybrid model that combined KNN and Naive Bayes algorithms. KNN is a type of machine learning algorithm that classifies a sample based on the classification of its nearest neighbors in the feature space. Naive Bayes is a probabilistic algorithm that makes predictions based on the probabilities of different events occurring.

The authors found that the hybrid model had an accuracy of 97.5% in classifying the samples as benign or malignant. This was slightly higher than the accuracies achieved by either KNN or Naive Bayes alone, which were 96.9% and 96.4%, respectively.

In conclusion, Al-Quaiz demonstrated that a hybrid approach combining KNN and Naive Bayes can be effective for diagnosing breast cancer from FNA samples, with high levels of accuracy. This has the potential to aid in the early detection and treatment of breast cancer, as it allows for the rapid and accurate diagnosis of breast cancer from FNA samples.

### K Nearest Neighbor (KNN)



***

The k-nearest neighbors algorithm is a non-parametric supervised learning method first developed by Evelyn Fix and Joseph Hodges in 1951, and later expanded by Thomas Cover. It is used for classification and regression. In both cases, the input consists of the k closest training examples in a data set. [28]

The algorithm uses proximity to make classifications or predictions about the grouping of an individual data point. [21]

It is commonly used for simple recommendation systems, pattern recognition, data mining, financial market predictions, intrusion detection, and more. [21]



### Support Vector Machine



***

Support vector machines (SVMs) are a set of supervised learning methods used for classification, regression and outliers detection. [17] 

Sklearn SVMs are commonly employed in classification tasks because they are particularly efficient in high-dimensional fields.

SVM's divide datasets into a large numbers of classes in order to discover a Maximum Marginal Hyperplane (MMH).  This is undertaken in two stages:

Support Vector Machines will initially iteratively build hyperplanes that best distinguish the classes.
It will then select the hyperplane that best separates the classes. [29]

### Naive Bayes



***

A Naive Bayes classifier is a probabilistic machine learning model that is used for classification task. The classifier is based on the Bayes theorem. 



"By using Bayes theorem, we can find the probability of A happening, given that B has occurred. Here, B is the evidence and A is the hypothesis. The assumption made here is that the predictors/features are independent. That is presence of one particular feature does not affect the other. Hence it is called naive." [29]

Naive Bayes uses a “naive” assumption of conditional independence between every pair of features given the value of the class variable. [18]

### What is Multicollinearity?

Multicollinearity occurs when two or more independent variables (also known as predictor) are highly correlated with one another in a regression model.

This means that an independent variable can be predicted from another independent variable in a regression model. 

```python
# Generate and visualize the correlation matrix [9]
corr = df.corr(numeric_only=True).round(2)

# Mask for the upper triangle
mask = np.zeros_like(corr, dtype=bool)
mask[np.triu_indices_from(mask)] = True

# Set figure size
f, ax = plt.subplots(figsize=(20, 20))

# Define custom colormap
cmap = sns.diverging_palette(220, 10, as_cmap=True)

# Draw the heatmap
sns.heatmap(corr, mask=mask, cmap=cmap, vmin=-1, vmax=1, center=0,
            square=True, linewidths=.5, cbar_kws={"shrink": .5}, annot=True)

plt.tight_layout()
```

We can verify the presence of multicollinearity between some of the variables.

### KNN - Nearest Neighbor

***

```python
# Making a Nearest Neighbor Classifier with Support Vector Machine [17] and Naive Bayes [18].
# Start with Models object and append the different classifiers to the object.

models = []

models.append(('KNN', KNeighborsClassifier()))
models.append(('NB', GaussianNB()))
models.append(('SVM', SVC()))
```

```python
# Cross Validate the Scores. 
# Use the re-sampling technique KFold, 
# which is useful for evaluating the machine learning nd dividing the Data Set into K parts.
# [19] [20]

from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import make_pipeline
```

```python
# Evaluating the Results

# Create 2 Arrays

results = []
names = []

for name, model in models:
    kfold=KFold(n_splits=10, random_state=None, shuffle=False)
    cv_results= cross_val_score(make_pipeline(StandardScaler(), model), X.loc[y_train.index], y_train, cv=kfold, scoring='accuracy')
    results.append(cv_results)
    names.append(name)

# Messages = Name, Mean and Standard Deviation

    msg = '%s:, %f:, (%f)' % (name, cv_results.mean(), cv_results.std())
    print(msg)
```

```python
# Since SVM is performing the best we will make a predictions on the test data set using it.

SVM = SVC()
SVM.fit(X_train, y_train)
predictions = SVM.predict(X_test)

print (accuracy_score(y_test, predictions))
print (classification_report(y_test, predictions))
print (confusion_matrix(y_test, predictions))
```
